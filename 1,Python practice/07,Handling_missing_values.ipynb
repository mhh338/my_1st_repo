# Handling Missing Values

Methods Of Handling Missing Values

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the dataset in pandas DataFrame (Download the 'placement-data.csv' file)
placement_df = pd.read_csv("/content/placement-data.csv")
```

```python
placement_df.head(100)
```

```python
placement_df.shape
```

```python
placement_df.isnull().sum()
```

```python
# Making some values as 'Nan'
target_rows = list(range(30,39)) + list(range(41,43)) + [55,59,61,62,65,68,72,75,78,81,85,90]
placement_df.loc[target_rows, 'cgpa'] = np.nan
```

1, Imputation Method,

(Replacing the missing values with statistical values known as central tendencies i.e., mean, median and mode).

```python
# Analyzing the distribution of data in the 'languages' column
sns.histplot(placement_df.cgpa)
```

i, Replacing The Missing Values Of 'languages' Column With Median

```python
# We get the skew-distribution, means, data is not evenly distributed,
# (data is distributed more on left side).
# So, we cannot use mean for replacing missing values.
placement_df["cgpa"].fillna(placement_df["cgpa"].median(), inplace=True)
```

```python
# Rechecking the missing values
placement_df.isnull().sum()
```

```python
# Plotting the distribution after replacing missing values.
sns.histplot(placement_df["cgpa"])
```

2, Dropping Method,

(Deleting all the rows having mising values in the corresponding column.)

```python
# Loading the dataset
cgpa_dataset = pd.read_csv("/content/placement-data.csv")
```

```python
# Shape of the dataset before dropping the values
cgpa_dataset.shape
```

```python
cgpa_dataset.head(100)
```

```python
# Making missing values,
target_rows = list(range(29,35)) + list(range(40,45)) + [49,53,55,56,60,61,68,70,78,82,88,95]
cgpa_dataset.loc[target_rows, 'cgpa'] = np.nan
```

```python
cgpa_dataset.isnull().sum()
```

```python
# Dropping the rows having '23' missing values
dropped_cgpa_dataset = cgpa_dataset.dropna(how = 'any')
```

```python
# Rechecking after the dropping the missing values
dropped_cgpa_dataset.isnull().sum()
```
