```python
# ============================================================
# PHASE 1B — GROUP BY PARENT STRUCTURE
# ============================================================

import json
import hashlib
import pandas as pd

def structure_signature(parent_structure):
    """
    Create a reproducible signature for the parent crystal.
    We ignore pymatgen metadata and retain the physical structure.
    """
    lattice = parent_structure["lattice"]["matrix"]

    sites = []
    for site in parent_structure["sites"]:
        element = site["species"][0]["element"]
        abc = site["abc"]
        sites.append({
            "element": element,
            "abc": [round(float(x), 6) for x in abc]
        })

    # Sort sites so ordering in the JSON does not matter
    sites = sorted(
        sites,
        key=lambda x: (x["element"], x["abc"][0], x["abc"][1], x["abc"][2])
    )

    obj = {
        "lattice": [
            [round(float(x), 6) for x in row]
            for row in lattice
        ],
        "sites": sites
    }

    text = json.dumps(obj, sort_keys=True)
    return hashlib.sha256(text.encode()).hexdigest()


# Build signatures
signatures = []

for r in data:
    signatures.append(structure_signature(r["parent_structure"]))

df["parent_structure_id"] = signatures

print("="*70)
print("PARENT STRUCTURE GROUPING")
print("="*70)

print("Total records:", len(df))
print("Unique parent structures:", df["parent_structure_id"].nunique())

counts = df.groupby("parent_structure_id").size()

print("\nRecords per parent structure:")
print(counts.describe())

print("\nParent structures with >1 magnetic ordering:",
      (counts > 1).sum())

print("Maximum orderings for one parent:",
      counts.max())
```

### Step - 1.7 - Extracting initial magnetic configuration

```python
# ============================================================
# PHASE 1F — EXTRACT INITIAL MAGNETIC CONFIGURATIONS
# ============================================================

import numpy as np
import pandas as pd


def extract_initial_magmoms(record):
    """
    Extract the initial magnetic moments from:
        record["input"]["structure"]["sites"][i]["properties"]["magmom"]

    Returns:
        list of floats
    """
    sites = record["input"]["structure"]["sites"]

    magmoms = []

    for site in sites:
        props = site.get("properties", {})
        magmom = props.get("magmom", 0.0)

        if magmom is None:
            magmom = 0.0

        magmoms.append(float(magmom))

    return magmoms


# ------------------------------------------------------------
# Extract every record
# ------------------------------------------------------------

rows = []

for r in data:

    initial_magmoms = extract_initial_magmoms(r)

    rows.append({
        "task_id": r.get("task_id"),
        "formula": r.get("formula_pretty"),
        "parent_structure_id": structure_signature(
            r["parent_structure"]
        ),

        "ordering": r.get("ordering"),
        "ordering_changed": r.get("ordering_changed"),

        "energy_per_atom": r.get("energy_per_atom"),
        "energy_above_ground": (
            r.get("energy_above_ground_state_per_atom")
        ),

        "initial_magmoms": initial_magmoms,

        "n_sites": len(initial_magmoms),

        "total_initial_moment": sum(initial_magmoms),

    })


df_mag = pd.DataFrame(rows)

print("=" * 80)
print("INITIAL MAGNETIC CONFIGURATION EXTRACTION")
print("=" * 80)

print("Total records:", len(df_mag))

print("\nMissing initial magnetic configurations:",
      df_mag["initial_magmoms"].isna().sum())

print("\nNumber of sites:")
print(df_mag["n_sites"].describe())

print("\nOrdering-changed:")
print(df_mag["ordering_changed"].value_counts(dropna=False))

print("\nExample:")
print(df_mag.iloc[0][[
    "formula",
    "ordering",
    "ordering_changed",
    "initial_magmoms",
    "total_initial_moment"
]])
```

### Step - 1.9 - Comparing initial vs final magnetic states

```python
# ============================================================
# PHASE 1H — COMPARE INITIAL VS FINAL MAGNETIC STATES
# ============================================================

def extract_final_magmoms(record):
    """
    Extract final VASP magnetic moments.
    """

    magmoms = record.get("magmoms", {})

    vasp = magmoms.get("vasp")

    if vasp is None:
        return None

    return [float(x) for x in vasp]


comparison = []

for r in data:

    initial = extract_initial_magmoms(r)
    final = extract_final_magmoms(r)

    if final is None:
        continue

    changed = False

    if len(initial) != len(final):
        changed = True

    else:
        changed = not np.allclose(
            initial,
            final,
            atol=0.1
        )

    comparison.append({
        "task_id": r.get("task_id"),
        "formula": r.get("formula_pretty"),
        "ordering": r.get("ordering"),
        "ordering_changed_database": r.get("ordering_changed"),
        "initial_magmoms": initial,
        "final_magmoms": final,
        "initial_final_different": changed
    })


df_compare = pd.DataFrame(comparison)

print("=" * 80)
print("INITIAL VS FINAL MAGNETIC STATES")
print("=" * 80)

print(
    df_compare["initial_final_different"]
    .value_counts(dropna=False)
)

print("\nDatabase ordering_changed:")
print(
    df_compare["ordering_changed_database"]
    .value_counts(dropna=False)
)
```

### Step - 1.10 - Assigning spin sign pattern

```python
def spin_sign_pattern(magmoms, threshold=0.1):

    pattern = []

    for m in magmoms:

        if m > threshold:
            pattern.append(+1)

        elif m < -threshold:
            pattern.append(-1)

        else:
            pattern.append(0)

    return tuple(pattern)
```

```python
df_mag["spin_pattern"] = df_mag["initial_magmoms"].apply(
    spin_sign_pattern
)
```

### Step - 1.14 - Determining the final magnetic configurations

Extracting the final signed moments and creating a final spin pattern.

```python
# ============================================================
# PHASE 1L — FINAL MAGNETIC CONFIGURATION
# ============================================================

def extract_final_magmoms(record):
    """
    Extract final VASP magnetic moments.
    """
    vasp = record.get("magmoms", {}).get("vasp")

    if vasp is None:
        return None

    return [float(x) for x in vasp]


def spin_sign_pattern(magmoms, threshold=0.1):
    """
    Convert magnetic moments into a simple collinear
    spin-sign representation.

    +1 = positive moment
    -1 = negative moment
     0 = approximately nonmagnetic
    """

    pattern = []

    for m in magmoms:

        if m > threshold:
            pattern.append(1)

        elif m < -threshold:
            pattern.append(-1)

        else:
            pattern.append(0)

    return tuple(pattern)


final_rows = []

for r in data:

    final_mags = extract_final_magmoms(r)

    if final_mags is None:
        continue

    final_rows.append({
        "task_id": r.get("task_id"),
        "formula": r.get("formula_pretty"),

        "parent_structure_id": structure_signature(
            r["parent_structure"]
        ),

        "ordering": r.get("ordering"),
        "ordering_changed": r.get("ordering_changed"),

        "energy_per_atom": r.get("energy_per_atom"),
        "energy_above_ground": (
            r.get("energy_above_ground_state_per_atom")
        ),

        "final_magmoms": final_mags,

        "final_spin_pattern": spin_sign_pattern(final_mags),

        "total_final_moment": sum(final_mags)
    })


df_final = pd.DataFrame(final_rows)

print("=" * 80)
print("FINAL MAGNETIC CONFIGURATION")
print("=" * 80)

print("Records:", len(df_final))

print("\nExample:")
print(
    df_final.iloc[0][[
        "formula",
        "ordering",
        "ordering_changed",
        "final_magmoms",
        "final_spin_pattern",
        "total_final_moment"
    ]]
)
```

### Step - 1.18 - Investigating identical pattern & different energies

```python
# ============================================================
# PHASE 1P — INVESTIGATE IDENTICAL PATTERN / DIFFERENT ENERGY
# ============================================================

# Pick the repeated state with the largest energy range
worst = pattern_frequency.sort_values(
    "energy_range",
    ascending=False
).iloc[0]

parent_id = worst["parent_structure_id"]
pattern = worst["final_spin_pattern"]

print("=" * 80)
print("WORST IDENTICAL-PATTERN ENERGY DISCREPANCY")
print("=" * 80)

print("Parent ID:", parent_id)
print("Final spin pattern:", pattern)
print("Energy range:", worst["energy_range"])

# Get the corresponding original records
matching_records = []

for r in data:

    pid = structure_signature(r["parent_structure"])

    final_mags = extract_final_magmoms(r)

    if final_mags is None:
        continue

    final_pattern = spin_sign_pattern(final_mags)

    if pid == parent_id and final_pattern == pattern:
        matching_records.append(r)

print("\nNumber of matching records:", len(matching_records))


# Print detailed information
for r in matching_records:

    initial = extract_initial_magmoms(r)
    final = extract_final_magmoms(r)

    print("\n" + "-" * 80)

    print("task_id:", r.get("task_id"))
    print("formula:", r.get("formula_pretty"))
    print("ordering:", r.get("ordering"))
    print("ordering_changed:", r.get("ordering_changed"))

    print("energy/atom:",
          r.get("energy_per_atom"))

    print("energy above ground:",
          r.get("energy_above_ground_state_per_atom"))

    print("\nInitial moments:")
    print(initial)

    print("\nFinal moments:")
    print(final)

    print("\nFinal spin pattern:")
    print(spin_sign_pattern(final))
```

### Step - 1.19 - Comparing final structures

### Step - 1.22 - Structure & spin sign

```python
# ============================================================
# PHASE 1T — STRUCTURE + SPIN SIGNATURE
# ============================================================

import json
import hashlib


def input_spin_structure_signature(record, threshold=0.1):

    structure = record["input"]["structure"]

    sites = []

    for site in structure["sites"]:

        element = site["species"][0]["element"]
        abc = site["abc"]

        m = float(
            site.get("properties", {})
                 .get("magmom", 0.0)
        )

        if m > threshold:
            spin = 1
        elif m < -threshold:
            spin = -1
        else:
            spin = 0

        sites.append({
            "element": element,
            "abc": [
                round(float(abc[0]), 6),
                round(float(abc[1]), 6),
                round(float(abc[2]), 6)
            ],
            "spin": spin
        })

    obj = {
        "lattice": [
            [
                round(float(x), 6)
                for x in row
            ]
            for row in structure["lattice"]["matrix"]
        ],
        "sites": sites
    }

    text = json.dumps(obj, sort_keys=True)

    return hashlib.sha256(
        text.encode()
    ).hexdigest()
```

```python
def input_magnetic_signature(record):
    """
    Signature based on input structure + actual initial magmom.
    """

    structure = record["input"]["structure"]

    sites = []

    for site in structure["sites"]:

        element = site["species"][0]["element"]

        abc = site["abc"]

        magmom = float(
            site.get("properties", {})
                .get("magmom", 0.0)
        )

        sites.append({
            "element": element,
            "abc": [
                round(float(abc[0]), 6),
                round(float(abc[1]), 6),
                round(float(abc[2]), 6)
            ],
            "magmom": round(magmom, 6)
        })

    obj = {
        "lattice": [
            [
                round(float(x), 6)
                for x in row
            ]
            for row in structure["lattice"]["matrix"]
        ],
        "sites": sites
    }

    text = json.dumps(obj, sort_keys=True)

    return hashlib.sha256(
        text.encode()
    ).hexdigest()
```

### Step - 2.3 - Making the correct train/validation/test split

Splitting the parents, not individual magnetic states

```python
# ============================================================
# PHASE 2C — PARENT-LEVEL SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

# Convert to a plain Python list
parents = (
    graph_df["parent_structure_id"]
    .dropna()
    .unique()
    .tolist()
)

print("Total parents:", len(parents))

# 80% train, 20% temporary
train_parents, temp_parents = train_test_split(
    parents,
    test_size=0.20,
    random_state=42
)

# 10% validation, 10% test
val_parents, test_parents = train_test_split(
    temp_parents,
    test_size=0.50,
    random_state=42
)

train_parents = set(train_parents)
val_parents = set(val_parents)
test_parents = set(test_parents)


# Assign split
def assign_split(parent_id):

    if parent_id in train_parents:
        return "train"

    elif parent_id in val_parents:
        return "val"

    elif parent_id in test_parents:
        return "test"

    else:
        raise ValueError(
            f"Parent {parent_id} was not assigned to a split."
        )


graph_df["split"] = (
    graph_df["parent_structure_id"]
    .apply(assign_split)
)


print("=" * 80)
print("PARENT-LEVEL SPLIT")
print("=" * 80)

print(
    graph_df.groupby("split")["parent_structure_id"]
    .nunique()
)

print("\nSamples:")
print(graph_df["split"].value_counts())

print("\nFinal counts:")

for split in ["train", "val", "test"]:

    subset = graph_df[
        graph_df["split"] == split
    ]

    print(
        f"{split:5s}: "
        f"{subset['parent_structure_id'].nunique()} parents, "
        f"{len(subset)} states"
    )
```

### Step - 2.10 - Evaluating ALIGNN baseline model

```python
import os
import torch
from torch.utils.data import DataLoader
from alignn.pure_lmdb_dataset import batch_torch_graph_pairs
from sklearn.metrics import mean_absolute_error, mean_squared_error

# ------------------------------------------------------------
# 1. Device
# ------------------------------------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)


# ------------------------------------------------------------
# 2. Test loader
# ------------------------------------------------------------

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0,
    collate_fn=lambda x: x,
    pin_memory=torch.cuda.is_available(),
)


# ------------------------------------------------------------
# 3. Recreate the same model architecture
# ------------------------------------------------------------

# Get atom feature dimension from one test sample
sample = test_dataset[0]
atom_feature_dim = sample[0].ndata["atom_features"].shape[1]

config = ALIGNNAtomWisePureConfig(
    name="alignn_atomwise_pure",
    alignn_layers=2,
    gcn_layers=2,
    atom_input_features=atom_feature_dim,
    edge_input_features=80,
    triplet_input_features=40,
    embedding_features=64,
    hidden_features=64,
    output_features=1,
    calculate_gradient=False,
    atomwise_output_features=0,
    graphwise_weight=1.0,
    gradwise_weight=0.0,
    stresswise_weight=0.0,
    atomwise_weight=0.0,
    energy_mult_natoms=False,
    use_penalty=False,
    classification=False,
)

model = ALIGNNAtomWisePure(config).to(device)


# ------------------------------------------------------------
# 4. Load BEST checkpoint
# ------------------------------------------------------------

model_path = (
    "/content/drive/MyDrive/"
    "Proj_2_Alt_Mags_Grnd_E/"
    "magnetic-topological-materials/"
    "models/alignn_baseline_best.pt"
)

checkpoint = torch.load(
    model_path,
    map_location=device,
)

model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print("Loaded best model from epoch:", checkpoint["epoch"])
print("Saved validation MAE:", checkpoint["val_mae"])


# ------------------------------------------------------------
# 5. Predict test set
# ------------------------------------------------------------

all_predictions = []
all_targets = []

with torch.no_grad():

    for batch in test_loader:

        g, lg = batch_torch_graph_pairs(
            [(item[0], item[1]) for item in batch]
        )

        target = torch.stack([
            torch.as_tensor(item[3], dtype=torch.float32)
            for item in batch
        ])

        g = g.to(device)
        lg = lg.to(device)
        target = target.to(device)

        result = model((g, lg, None))

        prediction = result["out"].reshape(-1)

        all_predictions.extend(
            prediction.cpu().numpy()
        )

        all_targets.extend(
            target.cpu().numpy()
        )


# ------------------------------------------------------------
# 6. Test regression metrics
# ------------------------------------------------------------

test_mae = mean_absolute_error(
    all_targets,
    all_predictions
)

test_rmse = mean_squared_error(
    all_targets,
    all_predictions
) ** 0.5

print("\n==============================")
print("TEST RESULTS")
print("==============================")
print(f"Test MAE  : {test_mae:.6f} eV/atom")
print(f"Test RMSE : {test_rmse:.6f} eV/atom")
```

ground-state identification accuracy.

We need to take all test predictions and, for each parent_structure_id:

$$ \hat{i}_{GS}=\arg\min_i \hat{\Delta E}_i $$

and compare that with:

$$ {i}_{GS}=\arg\min_i {\Delta E}_i $$

```python
import pandas as pd
import numpy as np

# Copy the test portion
test_df = graph_df[
    graph_df["split"] == "test"
].copy()

# Add predictions
test_df["pred_delta_E"] = np.array(all_predictions)

# ------------------------------------------------------------
# Find actual and predicted ground states for each parent
# ------------------------------------------------------------

actual_idx = (
    test_df.groupby("parent_structure_id")["delta_E"]
    .idxmin()
)

pred_idx = (
    test_df.groupby("parent_structure_id")["pred_delta_E"]
    .idxmin()
)

actual_gs = test_df.loc[
    actual_idx,
    ["parent_structure_id", "task_id", "delta_E"]
].copy()

pred_gs = test_df.loc[
    pred_idx,
    ["parent_structure_id", "task_id", "pred_delta_E"]
].copy()

actual_gs = actual_gs.rename(
    columns={
        "task_id": "actual_gs_task",
        "delta_E": "actual_gs_delta_E",
    }
)

pred_gs = pred_gs.rename(
    columns={
        "task_id": "pred_gs_task",
        "pred_delta_E": "pred_gs_delta_E",
    }
)

comparison = actual_gs.merge(
    pred_gs,
    on="parent_structure_id",
)

comparison["correct"] = (
    comparison["actual_gs_task"]
    == comparison["pred_gs_task"]
)

accuracy = comparison["correct"].mean()

print("Number of test parents:", len(comparison))
print(
    f"Ground-state identification accuracy: "
    f"{accuracy:.4%}"
)

print(
    "\nCorrect:",
    comparison["correct"].sum()
)

print(
    "Incorrect:",
    (~comparison["correct"]).sum()
)
```

The baseline model receives only the crystal structure:

$$ G\rightarrow \Delta E $$

But for a given parent material, the competing magnetic configurations have essentially the same crystal structure. They differ in their magnetic moments/configuration.

So the model sees:

$$ G_{\text{same}} \rightarrow ? $$

for:

FM, AFM, FiM

and has no information telling it which magnetic state it is evaluating.

That means the structure-only ALIGNN tends to give nearly identical predictions for different magnetic configurations.
So the 16.99% ground-state accuracy is our baseline/control result. It demonstrates why spin information is necessary.

| Model                 | Input                        |             Test MAE | Ground-state accuracy |
| --------------------- | ---------------------------- | -------------------: | --------------------: |
| Structure-only ALIGNN | Crystal structure            | **0.008624 eV/atom** |            **16.99%** |

### Step - 3.1 - Creating the spin-aware dataset wrappers

```python
import torch
from torch.utils.data import Dataset

class SpinAwareDataset(Dataset):

    def __init__(self, lmdb_dataset, dataframe):
        self.lmdb_dataset = lmdb_dataset
        self.df = dataframe.reset_index(drop=True).copy()

        assert len(self.lmdb_dataset) == len(self.df)

    def __len__(self):
        return len(self.lmdb_dataset)

    def __getitem__(self, idx):

        g, lg, lattice, label = self.lmdb_dataset[idx]

        magmoms = torch.tensor(
            self.df.iloc[idx]["initial_magmoms"],
            dtype=torch.float32
        ).reshape(-1, 1)

        if magmoms.shape[0] != g.num_nodes:
            raise ValueError(
                f"Index {idx}: {g.num_nodes} atoms but "
                f"{magmoms.shape[0]} magnetic moments"
            )

        # Make a copy so the original baseline graph is untouched.
        g.ndata["magmom"] = magmoms

        return g, lg, lattice, label


train_spin_dataset = SpinAwareDataset(
    train_dataset,
    graph_df[graph_df["split"] == "train"]
)

val_spin_dataset = SpinAwareDataset(
    val_dataset,
    graph_df[graph_df["split"] == "val"]
)

test_spin_dataset = SpinAwareDataset(
    test_dataset,
    graph_df[graph_df["split"] == "test"]
)

print("Train:", len(train_spin_dataset))
print("Val:", len(val_spin_dataset))
print("Test:", len(test_spin_dataset))
```

```python
# ============================================================
# TOP-K + ENERGY REGRET FOR SPIN-AWARE MODEL
# ============================================================

df = test_spin_df.copy()

# Rank predictions within each parent
df["pred_rank"] = (
    df.groupby("parent_structure_id")["pred_delta_E"]
      .rank(method="first", ascending=True)
)

# True ground-state task for each parent
true_gs_task = (
    df.loc[
        df.groupby("parent_structure_id")["delta_E"].idxmin(),
        ["parent_structure_id", "task_id"]
    ]
    .rename(columns={"task_id": "true_gs_task"})
)

# Model's top-1, top-3, top-5
top1 = (
    df[df["pred_rank"] == 1]
    [["parent_structure_id", "task_id", "delta_E"]]
    .rename(columns={
        "task_id": "selected_task",
        "delta_E": "regret"
    })
)

topk = top1.merge(
    true_gs_task,
    on="parent_structure_id"
)

top1_accuracy = (
    topk["selected_task"] == topk["true_gs_task"]
).mean()

# Top-3 and Top-5
top3 = (
    df[df["pred_rank"] <= 3]
    .groupby("parent_structure_id")
    .apply(
        lambda x: (
            x["task_id"].values
            == x["task_id"].values
        ).any()
    )
)

# More direct top-k calculation
true_tasks = (
    true_gs_task.set_index("parent_structure_id")["true_gs_task"]
)

def top_k_accuracy(k):
    correct = 0

    for parent, true_task in true_tasks.items():

        subset = df[
            (df["parent_structure_id"] == parent)
            & (df["pred_rank"] <= k)
        ]

        if true_task in subset["task_id"].values:
            correct += 1

    return correct / len(true_tasks)

top3_accuracy = top_k_accuracy(3)
top5_accuracy = top_k_accuracy(5)

print("Top-1 accuracy :", f"{top1_accuracy:.4%}")
print("Top-3 accuracy :", f"{top3_accuracy:.4%}")
print("Top-5 accuracy :", f"{top5_accuracy:.4%}")

print("\nEnergy regret:")
print("Mean :", topk["regret"].mean(), "eV/atom")
print("Median:", topk["regret"].median(), "eV/atom")
print("Max :", topk["regret"].max(), "eV/atom")

for threshold in [0.001, 0.002, 0.005, 0.010, 0.020]:
    fraction = (
        topk["regret"] < threshold
    ).mean()

    print(
        f"Regret < {threshold:.3f} eV/atom:",
        f"{fraction:.2%}"
    )
```

### Step - 4.5 - Spin-aware top-1 accuracy vs ground-state energy gap

```python
# ============================================================
# SPIN-AWARE TOP-1 ACCURACY VS GROUND-STATE ENERGY GAP
# ============================================================

df = test_spin_df.copy()

parent_results = []

for parent, group in df.groupby("parent_structure_id"):

    energies = np.sort(group["delta_E"].values)

    if len(energies) < 2:
        continue

    gap = energies[1] - energies[0]

    actual_task = group.loc[
        group["delta_E"].idxmin(),
        "task_id"
    ]

    predicted_task = group.loc[
        group["pred_delta_E"].idxmin(),
        "task_id"
    ]

    parent_results.append({
        "parent_structure_id": parent,
        "gap": gap,
        "correct": actual_task == predicted_task,
    })

gap_df = pd.DataFrame(parent_results)

bins = [
    0.0,
    0.001,
    0.005,
    0.010,
    np.inf,
]

labels = [
    "<1 meV",
    "1–5 meV",
    "5–10 meV",
    ">10 meV",
]

gap_df["gap_group"] = pd.cut(
    gap_df["gap"],
    bins=bins,
    labels=labels,
    right=False,
)

print(
    gap_df.groupby("gap_group", observed=False)
    .agg(
        parents=("correct", "size"),
        accuracy=("correct", "mean"),
    )
)
```

The key observation is:

$$ \boxed{\text{The smaller the energy gap, the harder it is to identify the exact ground state.}} $$

And 183 of 254 parents (72%) are in the hardest <1 meV/atom regime.

That explains a lot of the 24.7% overall Top-1 accuracy.

What this means for our project

Our model is not necessarily failing to learn magnetic physics. It is being asked to resolve differences like:

$$ 0.0000 \quad\text{vs}\quad 0.00024\ {\rm eV/atom} $$

while its typical error is around:

$$ 0.0083\ {\rm eV/atom}. $$

That's a difference of roughly 30× smaller than the test MAE.

We should change the training objective from pure absolute-energy regression to a ranking-aware objective.

Instead of only teaching:

$$ \hat{\Delta E}\approx\Delta E, $$

we explicitly teach:

$$ \boxed{ E_i<E_j \quad\Rightarrow\quad \hat E_i<\hat E_j } $$

for competing magnetic states from the same parent material.

That is much closer to what we actually care about.

A good next experiment is a combined loss:

$$ \boxed{ L=L_{\rm MSE}+\lambda L_{\rm rank} } $$

where the MSE preserves energy prediction while the ranking term encourages correct ordering.

### Step - 4.6 - Pairwise ranking accuracy

For two magnetic configurations of the same material, how often does the model correctly say which one has lower energy?

```python
# ============================================================
# SPIN-AWARE PAIRWISE RANKING ACCURACY
# ============================================================

df = test_spin_df.copy()

correct_pairs = 0
total_pairs = 0

for parent, group in df.groupby("parent_structure_id"):

    # Convert to arrays
    true_energy = group["delta_E"].to_numpy()
    pred_energy = group["pred_delta_E"].to_numpy()

    n = len(group)

    # Compare every unique pair
    for i in range(n):
        for j in range(i + 1, n):

            true_diff = true_energy[i] - true_energy[j]
            pred_diff = pred_energy[i] - pred_energy[j]

            # Ignore exact ties, if any
            if true_diff == 0:
                continue

            # Correct ordering if signs agree
            if true_diff * pred_diff > 0:
                correct_pairs += 1

            total_pairs += 1

pairwise_accuracy = correct_pairs / total_pairs

print("Total non-tied pairs:", total_pairs)
print("Correctly ranked pairs:", correct_pairs)
print(
    "Pairwise ranking accuracy:",
    f"{pairwise_accuracy:.4%}"
)
```

### Step - 4.7 - Pairwise ranking accuracy vs true energy seperation

```python
# ============================================================
# PAIRWISE RANKING ACCURACY VS TRUE ENERGY SEPARATION
# ============================================================

df = test_spin_df.copy()

bins = [
    0.0,
    0.001,
    0.005,
    0.010,
    float("inf"),
]

labels = [
    "<1 meV",
    "1–5 meV",
    "5–10 meV",
    ">10 meV",
]

stats = {
    label: {"correct": 0, "total": 0}
    for label in labels
}

for parent, group in df.groupby("parent_structure_id"):

    true_energy = group["delta_E"].to_numpy()
    pred_energy = group["pred_delta_E"].to_numpy()

    n = len(group)

    for i in range(n):
        for j in range(i + 1, n):

            true_diff = true_energy[i] - true_energy[j]

            if true_diff == 0:
                continue

            separation = abs(true_diff)

            pred_diff = pred_energy[i] - pred_energy[j]

            bin_index = np.digitize(
                separation,
                bins[1:-1],
                right=False
            )

            label = labels[bin_index]

            stats[label]["total"] += 1

            if true_diff * pred_diff > 0:
                stats[label]["correct"] += 1


print("Pairwise ranking accuracy by energy separation")
print("==============================================")

for label in labels:

    correct = stats[label]["correct"]
    total = stats[label]["total"]

    accuracy = (
        correct / total
        if total > 0
        else float("nan")
    )

    print(
        f"{label:8s} | "
        f"pairs = {total:5d} | "
        f"accuracy = {accuracy:.2%}"
    )
```

```python
import random
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader

train_spin_df = (
    graph_df[graph_df["split"] == "train"]
    .reset_index(drop=True)
)

val_spin_df = (
    graph_df[graph_df["split"] == "val"]
    .reset_index(drop=True)
)

def make_parent_batches(df, max_samples=32, shuffle=True):

    groups = []

    for _, group in df.groupby("parent_structure_id"):
        groups.append(group.index.tolist())

    if shuffle:
        random.shuffle(groups)

    batches = []
    current = []

    for group in groups:

        if current and len(current) + len(group) > max_samples:
            batches.append(current)
            current = []

        current.extend(group)

    if current:
        batches.append(current)

    return batches


train_batches = make_parent_batches(
    train_spin_df,
    max_samples=32,
    shuffle=True,
)

val_batches = make_parent_batches(
    val_spin_df,
    max_samples=32,
    shuffle=False,
)

print("Training batches:", len(train_batches))
print("Validation batches:", len(val_batches))
```

### Step - 5.4 - Ranking loss

```python
def parentwise_ranking_loss(
    prediction,
    target,
    parent_indices,
    min_gap=0.001,
    max_margin=0.010,
):
    """
    Pairwise ranking loss within each parent.

    Only pairs separated by >= 1 meV/atom
    are used for the ranking component.
    """

    losses = []

    unique_parents = torch.unique(parent_indices)

    for parent in unique_parents:

        idx = torch.where(
            parent_indices == parent
        )[0]

        if len(idx) < 2:
            continue

        y = target[idx]
        p = prediction[idx]

        for i in range(len(idx)):
            for j in range(i + 1, len(idx)):

                true_diff = y[j] - y[i]

                if torch.abs(true_diff) < min_gap:
                    continue

                pred_diff = p[j] - p[i]

                sign = torch.sign(true_diff)

                margin = torch.clamp(
                    torch.abs(true_diff),
                    max=max_margin,
                )

                violation = (
                    margin
                    - sign * pred_diff
                )

                losses.append(
                    F.relu(violation) ** 2
                )

    if not losses:
        return prediction.new_tensor(0.0)

    return torch.stack(losses).mean()
```

PT (inversion + time reversal) test

```python
# ------------------------------------------------------------
# Check whether any inversion + time reversal (PT)
# preserves the magnetic structure
# ------------------------------------------------------------

coord_tol = 1e-3
mag_tol = 0.05

pt_found = False

for op_number, op in enumerate(P_ops):

    all_sites_match = True

    for i, site in enumerate(pmg_structure):

        # Apply inversion operation in fractional coordinates
        transformed = op.operate(
            site.frac_coords
        )

        # Bring into [0,1)
        transformed = transformed % 1.0

        # Find matching crystallographic site
        distances = []

        for j, target_site in enumerate(
            pmg_structure
        ):

            diff = (
                transformed
                - target_site.frac_coords
            )

            # Minimum-image fractional difference
            diff -= np.round(diff)

            distances.append(
                np.linalg.norm(diff)
            )

        j = int(np.argmin(distances))

        if distances[j] > coord_tol:
            all_sites_match = False
            break

        # For inversion, an axial magnetic moment
        # is unchanged by P.
        #
        # Time reversal then changes:
        #
        #       m -> -m
        #
        # Therefore PT requires:
        #
        #       m_j = -m_i

        if abs(
            final_magmoms[j]
            + final_magmoms[i]
        ) > mag_tol:

            all_sites_match = False
            break

    if all_sites_match:
        pt_found = True
        print(
            "PT symmetry FOUND using inversion operation:",
            op_number
        )
        break


print("\nPT symmetry:", pt_found)

if pt_found:
    print(
        "This magnetic structure is NOT "
        "altermagnetic under the PT criterion."
    )
else:
    print(
        "PT is broken. Together with full spin "
        "compensation, this is an altermagnet candidate."
    )
```

### Step - 6.3 - Checking test parents as AM or not

```python
import numpy as np
import pandas as pd
from pymatgen.core import Structure
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

# ------------------------------------------------------------
# Map task_id -> original DFT record
# ------------------------------------------------------------

record_by_task = {
    str(r["task_id"]): r
    for r in data
}

# ------------------------------------------------------------
# PT checker
# ------------------------------------------------------------

def check_PT_symmetry(
    structure,
    magmoms,
    op,
    coord_tol=0.01,
    mag_tol=0.05,
):
    """
    Test whether inversion + time reversal preserves
    the collinear magnetic structure.

    For PT:
        position -> inversion(position)
        moment  -> -moment
    """

    for i, site in enumerate(structure):

        transformed = (
            op.operate(site.frac_coords) % 1.0
        )

        best_j = None
        best_dist = np.inf

        for j, target in enumerate(structure):

            diff = (
                transformed
                - target.frac_coords
            )

            # minimum-image fractional distance
            diff -= np.round(diff)

            dist = np.linalg.norm(diff)

            if dist < best_dist:
                best_dist = dist
                best_j = j

        if best_dist > coord_tol:
            return False

        # Under PT the magnetic moment changes sign
        if abs(
            magmoms[best_j] + magmoms[i]
        ) > mag_tol:
            return False

    return True


# ------------------------------------------------------------
# Altermagnet screening
# ------------------------------------------------------------

def screen_altermagnet_corrected(
    record,
    symprec=1e-2,
    angle_tolerance=5,
    net_tol=1e-4,
    coord_tol=0.01,
    mag_tol=0.05,
):
    """
    Preliminary altermagnet screen.

    Criterion:
        1. Fully compensated magnetic ground state
        2. Broken PT symmetry

    If inversion P is absent, PT is automatically absent
    because PT requires P.
    """

    structure = Structure.from_dict(
        record["structure"]
    )

    magmoms = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    net_moment = abs(magmoms.sum())

    compensated = net_moment <= net_tol

    if not compensated:
        return {
            "task_id": str(record["task_id"]),
            "formula": record["formula_pretty"],
            "ordering": record["ordering"],
            "net_moment": net_moment,
            "compensated": False,
            "spacegroup": None,
            "inversion_found": False,
            "PT_preserved": None,
            "AM_candidate": False,
        }

    sga = SpacegroupAnalyzer(
        structure,
        symprec=symprec,
        angle_tolerance=angle_tolerance,
    )

    ops = sga.get_symmetry_operations(
        cartesian=False
    )

    inversion_found = False
    pt_preserved = False

    # Search for an inversion operation.
    for op in ops:

        R = np.asarray(
            op.rotation_matrix,
            dtype=float
        )

        if np.allclose(
            R,
            -np.eye(3),
            atol=1e-5
        ):

            inversion_found = True

            if check_PT_symmetry(
                structure,
                magmoms,
                op,
                coord_tol=coord_tol,
                mag_tol=mag_tol,
            ):
                pt_preserved = True

            break

    # If P does not exist, PT cannot exist.
    if not inversion_found:
        pt_preserved = False

    # Correct AM screening criterion
    am_candidate = (
        compensated
        and not pt_preserved
    )

    return {
        "task_id": str(record["task_id"]),
        "formula": record["formula_pretty"],
        "ordering": record["ordering"],
        "net_moment": net_moment,
        "compensated": compensated,
        "spacegroup": sga.get_space_group_symbol(),
        "inversion_found": inversion_found,
        "PT_preserved": pt_preserved,
        "AM_candidate": am_candidate,
    }
```

Diagnostic

```python
# ============================================================
# ACTUAL MAGNETIC SYMMETRY: P / T / PT
# ============================================================

def magnetic_operation_preserves(
    structure,
    magmoms,
    op,
    time_reversal=+1,
    coord_tol=0.01,
    mag_tol=0.05,
):
    for i, site in enumerate(structure):

        transformed = (
            op.operate(site.frac_coords) % 1.0
        )

        best_j = None
        best_dist = np.inf

        for j, target in enumerate(structure):

            diff = transformed - target.frac_coords
            diff -= np.round(diff)

            dist = np.linalg.norm(diff)

            if dist < best_dist:
                best_dist = dist
                best_j = j

        if best_dist > coord_tol:
            return False

        expected = time_reversal * magmoms[i]

        if abs(
            magmoms[best_j] - expected
        ) > mag_tol:
            return False

    return True


rec = record_by_task["28294"]

s = Structure.from_dict(rec["structure"])

m = np.asarray(
    rec["magmoms"]["vasp"],
    dtype=float
)

sga = SpacegroupAnalyzer(
    s,
    symprec=1e-2,
    angle_tolerance=5
)

ops = sga.get_symmetry_operations(
    cartesian=False
)

print(
    "Crystallographic SG:",
    sga.get_space_group_symbol()
)

P_found = False
PT_found = False
T_found = False

identity = None

for op in ops:

    R = np.asarray(
        op.rotation_matrix,
        dtype=float
    )

    ordinary = magnetic_operation_preserves(
        s,
        m,
        op,
        time_reversal=+1,
    )

    with_T = magnetic_operation_preserves(
        s,
        m,
        op,
        time_reversal=-1,
    )

    # Identity operation
    if (
        np.allclose(R, np.eye(3), atol=1e-5)
        and np.allclose(
            op.translation_vector,
            [0, 0, 0],
            atol=1e-5,
        )
    ):
        identity = op

    # Inversion / inversion + T
    if np.allclose(
        R,
        -np.eye(3),
        atol=1e-5
    ):
        if ordinary:
            P_found = True

        if with_T:
            PT_found = True

    print(
        "\nRotation matrix:\n",
        R
    )

    print(
        "Translation:",
        op.translation_vector
    )

    print(
        "Preserves magnetic moments:",
        ordinary
    )

    print(
        "Preserves with time reversal:",
        with_T
    )


# Plain T = identity + time reversal
if identity is not None:

    T_found = magnetic_operation_preserves(
        s,
        m,
        identity,
        time_reversal=-1,
    )


print("\n==============================")
print("MAGNETIC SYMMETRY RESULT")
print("==============================")
print("P preserved :", P_found)
print("T preserved :", T_found)
print("PT preserved:", PT_found)
```

Applying this proper \(P/T/PT\) test to all 76 candidates by using the magnetic_operation_preserves() function

```python
# ============================================================
# RIGOROUS PRELIMINARY MAGNETIC-SYMMETRY SCREEN
# FOR ALL 76 COMPENSATED PT-BROKEN CANDIDATES
# ============================================================

def get_PTR_symmetry(record, symprec=0.01):

    structure = Structure.from_dict(
        record["structure"]
    )

    magmoms = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    sga = SpacegroupAnalyzer(
        structure,
        symprec=symprec,
        angle_tolerance=5,
    )

    ops = sga.get_symmetry_operations(
        cartesian=False
    )

    P_found = False
    PT_found = False
    T_found = False
    identity = None

    for op in ops:

        R = np.asarray(
            op.rotation_matrix,
            dtype=float
        )

        ordinary = magnetic_operation_preserves(
            structure,
            magmoms,
            op,
            time_reversal=+1,
        )

        with_T = magnetic_operation_preserves(
            structure,
            magmoms,
            op,
            time_reversal=-1,
        )

        if (
            np.allclose(R, np.eye(3), atol=1e-5)
            and np.allclose(
                op.translation_vector,
                [0, 0, 0],
                atol=1e-5,
            )
        ):
            identity = op

        if np.allclose(
            R,
            -np.eye(3),
            atol=1e-5
        ):
            if ordinary:
                P_found = True

            if with_T:
                PT_found = True

    if identity is not None:

        T_found = magnetic_operation_preserves(
            structure,
            magmoms,
            identity,
            time_reversal=-1,
        )

    return {
        "spacegroup": sga.get_space_group_symbol(),
        "P_preserved": P_found,
        "T_preserved": T_found,
        "PT_preserved": PT_found,
    }
```

```python
# ============================================================
# MAGNETIC SPACE GROUP DATA FOR ALL 55 CANDIDATES
# ============================================================

def get_magnetic_spacegroup_info(record, symprec=1e-2):
    structure = Structure.from_dict(record["structure"])

    elements = [
        site.species.elements[0].symbol
        for site in structure
    ]

    element_types = {
        el: i + 1
        for i, el in enumerate(sorted(set(elements)))
    }

    numbers = [
        element_types[el]
        for el in elements
    ]

    magmoms = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    cell = (
        np.asarray(structure.lattice.matrix, dtype=float),
        np.asarray(structure.frac_coords, dtype=float),
        np.asarray(numbers, dtype=int),
        magmoms,
    )

    mag_sym = spglib.get_magnetic_symmetry(
        cell,
        symprec=symprec,
        angle_tolerance=5,
        mag_symprec=0.05,
        is_axial=False,
        with_time_reversal=True,
    )

    if mag_sym is None:
        return {
            "task_id": str(record["task_id"]),
            "formula": record["formula_pretty"],
            "status": "magnetic_symmetry_not_found",
        }

    mst = spglib.get_magnetic_spacegroup_type_from_symmetry(
        mag_sym["rotations"],
        mag_sym["translations"],
        mag_sym["time_reversals"],
        lattice=cell[0],
    )

    # Spglib found operations, but could not assign an MSG type
    if mst is None:
        return {
            "task_id": str(record["task_id"]),
            "formula": record["formula_pretty"],
            "status": "msg_type_not_identified",
            "n_operations": len(mag_sym["rotations"]),
            "n_time_reversal_ops": int(
                np.sum(
                    np.asarray(
                        mag_sym["time_reversals"]
                    ) == 1
                )
            ),
        }

    return {
        "task_id": str(record["task_id"]),
        "formula": record["formula_pretty"],
        "status": "ok",
        "BNS_number": mst.bns_number,
        "OG_number": mst.og_number,
        "MSG_number": mst.number,
        "MSG_type": mst.type,
        "n_operations": len(
            mag_sym["rotations"]
        ),
        "n_time_reversal_ops": int(
            np.sum(
                np.asarray(
                    mag_sym["time_reversals"]
                ) == 1
            )
        ),
    }
```

What does “Top-K” mean?

For one material, suppose you have 8 possible magnetic configurations.

The true DFT energies might be:

| Configuration | DFT energy |
| ------------- | ---------: |
| A             |    -10.000 |
| B             |     -9.995 |
| C             |     -9.990 |
| D             |     -9.970 |
| E             |     -9.950 |
| F             |     -9.920 |
| G             |     -9.900 |
| H             |     -9.850 |

The true ground state is A because it has the lowest energy.

Your spin-aware ALIGNN does not know the DFT energies. It predicts energies and ranks the configurations.

Suppose it predicts this ranking:

C → B → A → D → E → F → G → H

Then:

* Top-1 = only C is selected.
* Top-2 = C and B are selected.
* Top-3 = C, B and A are selected.
* Top-5 = C, B, A, D and E are selected.

Because A is ranked 3rd, this material's ground state is recovered at Top-3, but not Top-1 or Top-2.

So:

Top-K means: take the K configurations that ALIGNN predicts to be lowest in energy, and then perform DFT on only those K configurations.

```python
import numpy as np
import pandas as pd
import spglib
from pymatgen.core import Element, Structure

# ------------------------------------------------------------
# Symmetry-based preliminary AM classifier
# ------------------------------------------------------------

def classify_preliminary_am(record, symprec=1e-2, net_tol=1e-3):

    structure_dict = record["input"]["structure"]

    structure = Structure.from_dict(structure_dict)

    lattice = np.asarray(structure.lattice.matrix, dtype=float)
    frac_coords = np.asarray(structure.frac_coords, dtype=float)

    numbers = np.array(
        [site.specie.Z for site in structure],
        dtype=int
    )

    magmoms = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    # Magnetic symmetry
    msym = spglib.get_magnetic_symmetry(
        (lattice, frac_coords, numbers, magmoms),
        symprec=symprec
    )

    if msym is None:
        return {
            "P_preserved": None,
            "T_preserved": None,
            "PT_preserved": None,
            "net_moment": float(np.sum(magmoms)),
            "compensated": False,
            "AM_candidate": False,
            "AM_type_preliminary": "unresolved",
        }

    rotations = np.asarray(msym["rotations"])
    translations = np.asarray(msym["translations"])
    time_reversals = np.asarray(msym["time_reversals"])

    I = np.eye(3, dtype=int)
    minus_I = -I

    # Pure inversion P
    P_preserved = any(
        np.array_equal(R, minus_I) and tr == 0
        for R, tr in zip(rotations, time_reversals * 0)
    )

    # The expression above cannot use translations that way;
    # use time_reversal explicitly below.
    P_preserved = any(
        np.array_equal(R, minus_I) and (tr == 0)
        and (int(trev) == 0)
        for R, tr, trev in zip(
            rotations, translations, time_reversals
        )
    )

    # Pure time reversal T:
    # spatial identity + time reversal
    T_preserved = any(
        np.array_equal(R, I) and (int(trev) == 1)
        for R, tr, trev in zip(
            rotations, translations, time_reversals
        )
        if np.allclose(tr % 1.0, 0.0, atol=symprec)
    )

    # PT:
    # inversion + time reversal
    PT_preserved = any(
        np.array_equal(R, minus_I) and (int(trev) == 1)
        for R, tr, trev in zip(
            rotations, translations, time_reversals
        )
    )

    net_moment = float(np.sum(magmoms))
    compensated = abs(net_moment) <= net_tol

    # Preliminary AM criterion:
    # compensated + T broken + PT broken
    AM_candidate = (
        compensated
        and not T_preserved
        and not PT_preserved
    )

    if AM_candidate:
        if P_preserved:
            am_type = "S-type"
        else:
            am_type = "S/A-type"
    else:
        am_type = "non-AM"

    return {
        "P_preserved": P_preserved,
        "T_preserved": T_preserved,
        "PT_preserved": PT_preserved,
        "net_moment": net_moment,
        "compensated": compensated,
        "AM_candidate": AM_candidate,
        "AM_type_preliminary": am_type,
    }
```

```python
def classify_preliminary_am(record, symprec=1e-2, net_tol=1e-3):

    structure = Structure.from_dict(
        record["input"]["structure"]
    )

    lattice = np.asarray(
        structure.lattice.matrix,
        dtype=float
    )

    frac_coords = np.asarray(
        structure.frac_coords,
        dtype=float
    )

    numbers = np.array(
        [site.specie.Z for site in structure],
        dtype=int
    )

    magmoms = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    msym = spglib.get_magnetic_symmetry(
        (lattice, frac_coords, numbers, magmoms),
        symprec=symprec
    )

    if msym is None:
        return {
            "P_preserved": None,
            "T_preserved": None,
            "PT_preserved": None,
            "net_moment": float(np.sum(magmoms)),
            "compensated": False,
            "AM_candidate": False,
            "AM_type_preliminary": "unresolved",
        }

    rotations = np.asarray(msym["rotations"])
    translations = np.asarray(msym["translations"])
    time_reversals = np.asarray(msym["time_reversals"])

    I = np.eye(3, dtype=int)
    minus_I = -I

    P_preserved = False
    T_preserved = False
    PT_preserved = False

    for R, tr, trev in zip(
        rotations,
        translations,
        time_reversals
    ):

        zero_translation = np.allclose(
            tr % 1.0,
            0.0,
            atol=symprec
        )

        if np.array_equal(R, minus_I) and trev == 0:
            P_preserved = True

        if (
            np.array_equal(R, I)
            and trev == 1
            and zero_translation
        ):
            T_preserved = True

        if np.array_equal(R, minus_I) and trev == 1:
            PT_preserved = True

    net_moment = float(np.sum(magmoms))

    compensated = abs(net_moment) <= net_tol

    AM_candidate = (
        compensated
        and not T_preserved
        and not PT_preserved
    )

    if AM_candidate:
        AM_type = (
            "S-type"
            if P_preserved
            else "S/A-type"
        )
    else:
        AM_type = "non-AM"

    return {
        "P_preserved": P_preserved,
        "T_preserved": T_preserved,
        "PT_preserved": PT_preserved,
        "net_moment": net_moment,
        "compensated": compensated,
        "AM_candidate": AM_candidate,
        "AM_type_preliminary": AM_type,
    }
```

```python
# ============================================================
# ERROR ANALYSIS: THE 3 AM STATES MISSED BY TOP-10
# ============================================================

missed_top10 = am_rank_typed[
    am_rank_typed["AM_ground_rank"] > 10
].copy()

print("Missed AM states:")
print(
    missed_top10[
        [
            "task_id",
            "formula",
            "AM_type_preliminary",
            "AM_ground_rank",
            "n_configs",
        ]
    ].to_string(index=False)
)


# ------------------------------------------------------------
# Detailed analysis for each missed AM ground state
# ------------------------------------------------------------

error_rows = []

for _, row in missed_top10.iterrows():

    task_id = row["task_id"]
    parent_id = row["parent_structure_id"]

    group = (
        test_df[
            test_df["parent_structure_id"] == parent_id
        ]
        .copy()
        .sort_values("pred_energy")
        .reset_index(drop=True)
    )

    # Actual ground state
    actual_row = group.loc[
        group["task_id"] == task_id
    ].iloc[0]

    # ML-selected state
    predicted_row = group.iloc[0]

    # Actual first-excited-state gap
    true_sorted = group.sort_values("delta_E")

    energies = true_sorted["delta_E"].to_numpy()

    if len(energies) >= 2:
        true_gap = energies[1] - energies[0]
    else:
        true_gap = np.nan

    # Absolute prediction error of actual AM ground state
    actual_prediction_error = (
        abs(
            actual_row["pred_energy"]
            - actual_row["delta_E"]
        )
    )

    # Energy regret of ML-selected state
    regret = predicted_row["delta_E"]

    error_rows.append({
        "task_id": task_id,
        "formula": row["formula"],
        "AM_type": row["AM_type_preliminary"],
        "AM_rank": row["AM_ground_rank"],
        "n_configs": len(group),
        "true_gap_eV_atom": true_gap,
        "AM_true_delta_E": actual_row["delta_E"],
        "AM_predicted_delta_E": actual_row["pred_energy"],
        "AM_abs_prediction_error":
            actual_prediction_error,
        "ML_selected_task":
            predicted_row["task_id"],
        "ML_selected_true_delta_E":
            predicted_row["delta_E"],
        "ML_selected_regret":
            regret,
    })

error_df = pd.DataFrame(error_rows)

print("\n==============================")
print("TOP-10 MISSED AM ERROR ANALYSIS")
print("==============================")

print(
    error_df.to_string(index=False)
)
```

### Step - 9.4 - Validating the symmetry function on task 28294 again using the actual final DFT structure and final VASP moments.

```python
# ============================================================
# DEFINE MAGNETIC P / T / PT SYMMETRY TEST
# ============================================================

import spglib
import numpy as np


def magnetic_PT_classification(
    structure,
    magmoms,
    symprec=1e-2,
    mag_symprec=1e-2,
):

    lattice = np.array(
        structure.lattice.matrix,
        dtype=float
    )

    positions = np.array(
        structure.frac_coords,
        dtype=float
    )

    numbers = np.array(
        structure.atomic_numbers,
        dtype=int
    )

    magmoms = np.array(
        magmoms,
        dtype=float
    )

    # spglib magnetic cell
    cell = (
        lattice,
        positions,
        numbers,
        magmoms
    )

    mag_sym = spglib.get_magnetic_symmetry(
        cell,
        symprec=symprec,
        mag_symprec=mag_symprec
    )

    if mag_sym is None:
        return {
            "P_preserved": np.nan,
            "T_preserved": np.nan,
            "PT_preserved": np.nan,
            "n_operations": 0,
            "n_time_reversal_ops": 0,
            "symmetry_status": "failed",
        }

    rotations = np.asarray(
        mag_sym["rotations"],
        dtype=int
    )

    time_reversals = np.asarray(
        mag_sym["time_reversals"],
        dtype=bool
    )

    I = np.eye(3, dtype=int)
    minus_I = -I

    # P = spatial inversion only
    P_preserved = np.any(
        np.all(rotations == minus_I, axis=(1, 2))
        & (~time_reversals)
    )

    # T = time reversal only
    T_preserved = np.any(
        np.all(rotations == I, axis=(1, 2))
        & time_reversals
    )

    # PT = inversion + time reversal
    PT_preserved = np.any(
        np.all(rotations == minus_I, axis=(1, 2))
        & time_reversals
    )

    return {
        "P_preserved": bool(P_preserved),
        "T_preserved": bool(T_preserved),
        "PT_preserved": bool(PT_preserved),
        "n_operations": len(rotations),
        "n_time_reversal_ops": int(time_reversals.sum()),
        "symmetry_status": "ok",
    }

print("magnetic_PT_classification() is ready.")
```

```python
# ============================================================
# STEP 20: SPIN-SPACE ALTERMAGNET TEST
# ============================================================

import numpy as np
import pandas as pd
from pymatgen.core import Structure
from spinspg import get_spin_symmetry


SPIN_TOL = 0.05


def spin_space_am_test(
    record,
    symprec=1e-2,
    mag_symprec=5e-2,
    spin_tol=0.05,
):
    """
    Preliminary spin-space test for collinear altermagnetism.

    We look for a spin-space operation where:
        spatial operation maps one magnetic sublattice
        to the opposite-spin sublattice,

    i.e. the spin rotation sends the collinear axis
    n -> -n.

    Pure translation (R = I) and pure inversion (R = -I)
    are excluded from the AM-generating operation.
    """

    structure = Structure.from_dict(record["structure"])

    raw_m = np.asarray(
        record["magmoms"]["vasp"],
        dtype=float
    )

    # Remove tiny induced moments so that weak polarization
    # on ligand / nonmagnetic atoms does not artificially
    # destroy the collinear symmetry.
    m = raw_m.copy()
    m[np.abs(m) < spin_tol] = 0.0

    # Collinear moments along a common axis.
    magmoms_3d = np.zeros((len(m), 3))
    magmoms_3d[:, 2] = m

    lattice = np.asarray(
        structure.lattice.matrix,
        dtype=float
    )

    positions = np.asarray(
        structure.frac_coords,
        dtype=float
    )

    numbers = np.asarray(
        structure.atomic_numbers,
        dtype=int
    )

    sog, rotations, translations, spin_rotations = (
        get_spin_symmetry(
            lattice,
            positions,
            numbers,
            magmoms_3d,
            symprec=symprec,
            mag_symprec=mag_symprec,
            angle_tolerance=5,
        )
    )

    axis = np.array([0.0, 0.0, 1.0])

    spin_flip_ops = []

    for i, (R, t, W) in enumerate(
        zip(
            np.asarray(rotations),
            np.asarray(translations),
            np.asarray(spin_rotations),
        )
    ):

        R = np.asarray(R, dtype=float)
        t = np.asarray(t, dtype=float)
        W = np.asarray(W, dtype=float)

        # What does the spin rotation do to the collinear axis?
        transformed_axis = W @ axis

        flips_spin = np.allclose(
            transformed_axis,
            -axis,
            atol=1e-5
        )

        if not flips_spin:
            continue

        # Exclude pure translation
        pure_translation = np.allclose(
            R,
            np.eye(3),
            atol=1e-5
        )

        # Exclude pure inversion / inversion + translation
        pure_inversion = np.allclose(
            R,
            -np.eye(3),
            atol=1e-5
        )

        if pure_translation or pure_inversion:
            continue

        spin_flip_ops.append({
            "operation_index": i,
            "R": R,
            "translation": t,
            "W": W,
            "det_R": np.linalg.det(R),
        })

    return {
        "spin_group": str(sog),
        "n_spin_space_operations": len(rotations),
        "n_spin_flip_rotation_ops": len(spin_flip_ops),
        "spin_flip_ops": spin_flip_ops,
        "spin_space_AM_candidate": (
            len(spin_flip_ops) > 0
        ),
        "cleaned_magmoms": m,
    }
```

Classifying the spatial operation

For each \(R\), we want to distinguish:

* proper rotation: \(\det R=+1\)
* improper rotation/mirror: \(\det R=-1\)
* identity
* inversion

```python
# ============================================================
# CLASSIFY SPATIAL OPERATIONS
# ============================================================

def classify_rotation_matrix(R, tol=1e-5):

    R = np.asarray(R, dtype=float)

    I = np.eye(3)
    inv = -I

    if np.allclose(R, I, atol=tol):
        return "identity"

    if np.allclose(R, inv, atol=tol):
        return "inversion"

    det = np.linalg.det(R)

    if det > 0:
        return "proper rotation"

    return "improper rotation / mirror"


spin_ops_df["spatial_operation_type"] = (
    spin_ops_df["R"]
    .apply(classify_rotation_matrix)
)

print(
    spin_ops_df[
        "spatial_operation_type"
    ].value_counts()
)
```

4. What does "spin-space" mean?

This is the extra symmetry analysis we introduced after the 70.

Ordinary magnetic-space symmetry looks mainly at:

$$ \text{position/space} + \text{time reversal}. $$

A spin-space symmetry goes one step further and considers a symmetry operation acting in real space and spin space together.

Very roughly:

$$ \text{spatial operation }R + \text{spin rotation }W. $$

For a collinear magnetic structure, a useful example is:

$$ \boxed{ R:\text{ maps one magnetic sublattice to another} } $$

while

$$ \boxed{ W:\uparrow\rightarrow\downarrow. } $$

So instead of merely asking:

"Is \(PT\) broken?"

we ask:

"Is there a nontrivial spatial operation that connects opposite-spin sublattices?"

That is much closer to the symmetry mechanism associated with altermagnetism.

For the 22 candidates, we need to test whether the electronic structure actually shows the characteristic altermagnetic spin splitting.

The workflow becomes:

$$\boxed{\text{22 candidates → DFT electronic structure → SOC OFF → spin-resolved bands →}ΔE_{↑↓}(k)}$$

We have 22 candidates. We don't need to do the most expensive calculation for all 22 immediately.

Lets follow this order:

$$ \boxed{\text{22}} \rightarrow \text{select 5–6 strongest candidates} \rightarrow \text{SOC-off band structures} $$

and then, depending on the results, expand to the remaining candidates.

The strongest starting candidates from your current list are:

* Li₃SiNi₃O₈
* LiVSiO₄
* CoH₅CN(ClO)₂
* Ba₂FeWO₆
* Co(IO₃)₂
* Mn₃TeO₆

These have the largest ground-to-first-excited magnetic energy gaps within the 22.

The important point is that symmetry has given us 22 candidates, but the decisive physical signature is non-relativistic spin splitting (NRSS): spin-up/down bands can be split even with SOC turned off. Strong altermagnets have this zero-SOC splitting; weak ones require SOC.

**What the DFT validation must show**

For each candidate we want:

$$ M\simeq0 $$ $$ PT\text{ broken} $$ $$ \text{nontrivial spin-space symmetry} $$

and finally

$$ \boxed{\Delta E_{\uparrow\downarrow}(\mathbf k)\neq0 \quad\text{with SOC OFF}} $$

at least over appropriate regions of the Brillouin zone.

The splitting does not have to occur everywhere: altermagnetic symmetry can enforce degeneracies on certain symmetry planes/lines while allowing splitting elsewhere. Recent calculations demonstrate exactly this behavior.

$$\boxed{\text{DFT ground state → SOC OFF → spin splitting? → YES → strong-AM candidate → SOC ON → check how SOC modifies splitting}}$$

For candidates that do not split without SOC, an SOC calculation can tell us whether they are potentially weak altermagnets.

### Step - 11.2 - Generating the exact inputs from the actual structure

This will produce the POSCAR and a first-pass INCAR/MAGMOM

```python
# ============================================================
# PREPARE Li3SiNi3O8 (task 34911) FOR NON-SOC DFT
# ============================================================

from pymatgen.core import Structure
from pathlib import Path

task_id = 34911
rec = record_by_task[str(task_id)]

structure = Structure.from_dict(rec["structure"])

# Final moments from the previous DFT calculation
final_m = np.asarray(
    rec["magmoms"]["vasp"],
    dtype=float
)

print("Number of sites:", len(structure))
print("Elements:", [str(site.specie) for site in structure])
print("Final magnetic moments:")
print(final_m)

# Write POSCAR
outdir = Path("/content/Li3SiNi3O8_task34911")
outdir.mkdir(exist_ok=True)

structure.to(
    fmt="poscar",
    filename=str(outdir / "POSCAR")
)

# Construct a simple initial MAGMOM:
# use the sign of the converged moments, with ~1.2 μB
# on the Ni sites and zero on nominally nonmagnetic sites.
initial_magmom = []

for site, m in zip(structure, final_m):

    element = str(site.specie)

    if element == "Ni" and abs(m) > 0.05:
        initial_magmom.append(
            1.2 if m > 0 else -1.2
        )
    else:
        initial_magmom.append(0.0)

print("\nInitial MAGMOM:")
print(initial_magmom)

print("\nVASP MAGMOM line:")
print(
    "MAGMOM = "
    + " ".join(f"{x:.1f}" for x in initial_magmom)
)

print("\nFiles written to:")
print(outdir)
```
