# Label Encoding

Converting The Labels To Numeric Form

```python
# Importing necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

# 1, Label Encoding Of Breast Cancer Dataset

```python
# Loading dataset to pandas DataFrame
bc_df = pd.read_csv("/content/bc_data.csv")
```

```python
bc_df.head()
```

```python
bc_df.shape
```

```python
# Finding count of labels in a feature
bc_df["diagnosis"].value_counts()
```

Converting The Corresponding Labels Into Numerical Values

```python
# Loading label encoder function
label_enc = LabelEncoder()
# Transforming the labels
bc_labels = label_enc.fit_transform(bc_df.diagnosis)
# Adding new column 'target'
bc_df["target"] = bc_labels
```

```python
bc_df.head()
```

0 = Benign (B)

1 = Malignant (M)

```python
# Counting values
bc_df["target"].value_counts()
```
