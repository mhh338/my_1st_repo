```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

Data Pre-Processing

```python
# Loading the data into pandas dataframe
diab_df = pd.read_csv("/content/diabetes_data.csv")
```

```python
diab_df.head()
```

```python
diab_df.shape
```

```python
# Getting the statistical measures of the data
diab_df.describe()
```

```python
# Counting values of 'Outcome' feature
diab_df["Outcome"].value_counts()
```

0 = Non-Diabetic

1 = Diabetic

Grouping the 'Outcome' feature by mean

```python
# The difference in themean values of two features is useful for model to predict correct values.
diab_df.groupby("Outcome").mean()
```

# Seperating The Data Into Features & Targets/Labels

```python
X = diab_df.drop(columns = "Outcome", axis = 1)
Y = diab_df["Outcome"]
```

```python
print(X, "\n\n\n", Y)
```

# Standardizing The Data

To make the data values lie within same range

```python
# Checking standard deviation of the data
X.std()
```

```python
# Loading the standard scaler function
scaler = StandardScaler()
scaler.fit(X)
X_standardized = scaler.transform(X)
```

```python
print(X_standardized)
```

```python
# Rechecking the standard deviation
X_standardized.std()
```

Splitting The Data Into Training & Testing Data

```python
""" 'stratify = Y' ensures that the training and testing data maintain the exact same proportion of class labels as of original dataset. """
X_train, X_test, Y_train, Y_test = train_test_split(X_standardized, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

```python
print(X_standardized.shape, X_train.shape, X_test.shape)
```

# Training The Model

```python
# Loading the 'Support Vector Classifier (SVC)'
classifier = svm.SVC(kernel ="linear")
```

```python
# Training the svm classifier
classifier.fit(X_train, Y_train)
```

# Evaluating The Model

1, Accuracy Score On The Training Data

```python
X_train_pred = classifier.predict(X_train)
train_data_acc = accuracy_score(X_train_pred, Y_train)
```

```python
print("Accuracy Score On The Training Data: ", train_data_acc * 100, "%")
```

2, Accuracy Score On The Testing Data

```python
X_test_pred = classifier.predict(X_test)
test_data_acc = accuracy_score(X_test_pred, Y_test)
```

```python
print("Accuracy On The Testing Data: ", test_data_acc * 100, "%")
```

# Making A Predictive System


```python
# Making an instance
input_data = (4,110,92,0,0,37.6,0.191,30)

# Converting 'input_data' to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshaping the array as we are predicting only one instance, and the model is trained on (768, 8) data points.
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# We have to standardize the data before feeding it to the model as we did that before
std_data = scaler.transform(input_data_reshaped) # As we fitted the 'X' before

# Predicting the result
prediction = classifier.predict(std_data)
print("The Predicted Result Is: ", prediction)

if(prediction[0] == 0):
  print("The Person Is Non-Diabetic")
else:
  print("The Person Is Diabetic")
```
