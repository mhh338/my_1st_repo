# Train Test Split Function

```python
# Importing necessary libraries
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

Download the dataset using following link,

```python
# Loading the dataset to pandas DataFrame
db_df = pd.read_csv("/content/diab_data.csv")
```

Splitting The Data Into Features & Targets/Labels

```python
# The features
X = db_df.drop(columns = "Outcome", axis = 1)
# The targets/labels
Y = db_df["Outcome"]
```

# Standardizing The Data

To Make The Data In Common Range

```python
# Checking standard deviation in X
X.std()
```

```python
# Using StandardScaler function for standardizing
scaler = StandardScaler()
scaler.fit(X)
stndrd_data = scaler.transform(X)
```

```python
# Checking standard deviation in standardized X data
stndrd_data.std()
```

```python
X = stndrd_data
Y = db_df["Outcome"]
```

# Splitting The Data Into Training & Testing Data

```python
# Using train test split function for splitting
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 2)
# 'test_size = 0.2' mean 20% of data will be test data and 80% will be the train data.
```
