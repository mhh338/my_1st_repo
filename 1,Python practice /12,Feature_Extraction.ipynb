# Feature Extraction Using TF-IDF Vectorizer

The Mapping From Textual Data To Real Valued Vectors Is Called Feature Extraction.

TF-IDF (Term Frequency - Inverse Document Frequency),

To Count Number Of Times Each Word Appears In A Document.

i, Term Frequency = (Number of times term 't' appears in a document/ Number of terms in a document) OR

TF = nt/T (n = no. of times, T = total terms),

ii, Inverse Document Frequency = (log of number of documents/log of number of documents in which term 't' has appeared in.) OR

IDF = log(N/n) (N = total number of documents, n = number of documents in which term 't' is appeared in it.)

iii, IDF value for rare word is high,
Whereas, for frequent word its value is low.

iv, At the end both values are miltiplied,

   TF-IDF value of t = TF × IDF

```python
# Merging the 'author' & 'title' features
train_df['content'] = train_df['author'] + ' ' + train_df['title']
```

# Vectorizing The Feature Text Using TF-IDF Vectorizer

Converting The Textual Data Into Feature Vectors

```python
# Loading the vectorizer function
vectorizer = TfidfVectorizer()

# Ensure all elements in X are strings and handle potential np.nan values
# TfidfVectorizer expects string type, so convert any non-string (like np.nan) to an empty string
import pandas as pd # Import pandas to use pd.notna
X_processed = np.array([str(item) if pd.notna(item) else "" for item in X], dtype=object)

vectorizer.fit(X_processed)
X_vectorized = vectorizer.transform(X_processed)
```

'X_vectorized' contain TF-IDF values for each individual term/word in the 'content' feature, this can then be fed to ML model.
