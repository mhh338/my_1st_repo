# Undersampling, Technique To Handle Imbalance Data

Splitting The Legit & Fraud Transactions Data

```python
legit = cred_card_df[cred_card_df.Class == 0]
fraud = cred_card_df[cred_card_df.Class == 1]
```

```python
print(f"Legit Shape:{legit.shape}\nFraud Shape:{fraud.shape}")
```

Building A Sample Dataset Containing Similar Distribution Of Legit & Fraud Transactions

Number Of Fraudulent Transaction = 85

Taking Similar Amount From Legit Transactions

```python
# Using 'sample' function
legit_sample = legit.sample(n=85)
```

```python
legit_sample.shape
```

# Concatenating Legit-sample & Fraud DataFrames

```python
# 'axis = 0' means we want to concatenate row-wise, for column wise 'axis = 1'
upd_cred_card_df = pd.concat([legit_sample, fraud], axis = 0)
```
